```python
# Example of execute()
# Used for DDL and executing a single SQL statement.

import sqlite3

# Connect to SQLite database
sqlite_conn = sqlite3.connect('keys.db')

try:
    sqlite_cursor = sqlite_conn.cursor()

    # Create table
    sqlite_cursor.execute('''
        CREATE TABLE IF NOT EXISTS keys (
            id INTEGER PRIMARY KEY,
            bucketname TEXT NOT NULL,
            accesskey TEXT NOT NULL,
            secretkey TEXT NOT NULL
        );
    ''')

    # delete Data
    sqlite_cursor.execute(
        '''
        Delete from keys
        ''' )
    
    # Insert data
    sqlite_cursor.execute(
        '''
        INSERT INTO keys (bucketname, accesskey, secretkey)
        VALUES (?, ?, ?)
        ''',
        (
            "xx",
            "yy",
            "zz"
        )
    )

    # Commit changes
    sqlite_conn.commit()

    print("Record inserted successfully")
    
finally:
    sqlite_conn.close()
```

```python
# fetchone example

import sqlite3

sqlite_conn = sqlite3.connect('keys.db')
sqlite_cursor = sqlite_conn.cursor()

# Using fetchone() to get results one by one
sqlite_cursor.execute('''SELECT bucketname, accesskey, secretkey FROM keys;''')
print("Key Values  database:")
row = sqlite_cursor.fetchone()  # Fetch one row at a time
print(row)
bucketname,keyname,secretkey = row

print(bucketname)
print(keyname)
print(secretkey)

sqlite_cursor.close()
sqlite_conn.commit()
sqlite_conn.close()
```
